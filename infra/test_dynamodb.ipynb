```python
import boto3
from boto3.dynamodb.conditions import Key

dynamodb = boto3.resource(
    "dynamodb",
    endpoint_url="http://localhost:8000",
    region_name="us-east-1",
    aws_access_key_id="fake",
    aws_secret_access_key="fake",
)
table = dynamodb.Table("mediguess-scores")
```

```python
table.put_item(Item={"player_id": "nina", "played_at": "2026-09-25T10:00:00Z", "score": 42, "board": "GLOBAL"})
table.put_item(Item={"player_id": "paul", "played_at": "2026-09-25T10:05:00Z", "score": 57, "board": "GLOBAL"})
```

```python
resp = table.query(
    IndexName="by_score",
    KeyConditionExpression=Key("board").eq("GLOBAL"),
    ScanIndexForward=False,
    Limit=10,
)
for item in resp["Items"]:
    print(item["player_id"], item["score"])
```
